## 1. What is the core business problem and why does it matter?

The core business problem is managing bar inventory efficiently while ensuring that sufficient stock is available to meet customer demand.

Hotels and bars need to maintain the right quantity of beverages such as beer, wine, vodka, rum, and whiskey. Overstocking increases storage costs, wastage, and blocked capital, while understocking can lead to stockouts, customer dissatisfaction, and lost sales.

My solution uses historical inventory consumption data to:

* Forecast future demand for individual items.

* Recommend inventory par levels.

* Support better purchasing and replenishment decisions.

* Reduce the risk of stockouts and unnecessary overstocking.

## 2. What assumptions did you make? Why?

I made the following assumptions:

1. Historical consumption is a useful indicator of future demand. The model uses past consumption patterns to estimate upcoming requirements.

2. The data represents regular inventory operations. I assumed that the recorded consumption, purchases, and closing inventory values were reasonably accurate.

3. A three-day lead time is required for replenishment. This represents the assumed time between placing an order and receiving stock.

4. The hotel reviews inventory requirements over a seven-day period. This helps estimate stock requirements beyond immediate daily demand.

5. Demand variability should be considered. I included safety stock to reduce the risk of stockouts caused by unexpected demand.

6. Historical patterns may not capture special events or seasonal changes. Therefore, the system is treated as a baseline forecasting solution that can be improved with additional business information.

These assumptions were necessary because the dataset did not provide details such as supplier lead times, promotions, holidays, or special hotel events.

## 3. What model did you use and why did you choose it? Why not others?

I used a moving-average-based demand forecasting approach with safety-stock-based par level recommendations.

The system aggregates historical consumption and uses recent demand patterns to estimate future consumption. It then calculates recommended inventory levels using forecast demand, lead time, review period, and safety stock.

### Why I chose this approach

* It is simple and easy to explain to hotel inventory managers.

* It provides a useful baseline for demand forecasting.

* It works with the available historical consumption data.

* It requires fewer assumptions than complex forecasting models.

* It is computationally efficient and easy to integrate into a dashboard.

### Why not other models?

I did not initially choose advanced models such as XGBoost, LSTM, or Prophet because:

* The project required a practical and explainable solution.

* Additional feature engineering would be needed for advanced models.

* The dataset did not include important external factors such as events, promotions, and holidays.

* A baseline model should be established before comparing more complex models.

In the future, I would compare the baseline with models such as Random Forest, XGBoost, Prophet, and time-series models to determine whether they provide measurable improvements.

## 4. How does your system perform? What would you improve?

The baseline system was evaluated using historical consumption data.

* Mean Absolute Error (MAE): Approximately 92.30 ml

* WAPE: Approximately 165.98%

The MAE indicates the average absolute difference between the predicted and actual consumption values. However, the high WAPE indicates that the baseline has limitations, particularly for items or periods with low consumption.

### Improvements I would make

1. Compare multiple forecasting models instead of relying only on the moving average.

2. Use item-level and bar-level forecasting.

3. Add day-of-week, month, holiday, and seasonal features.

4. Include special events, promotions, and occupancy information.

5. Use time-series cross-validation.

6. Evaluate performance using MAE, RMSE, WAPE, and stockout-related business metrics.

7. Tune safety stock and service-level parameters based on actual hotel requirements.

8. Monitor model performance continuously after deployment.

The current system should therefore be considered a baseline prototype, not a fully optimized production forecasting system.

## 5. How would this solution work in a real hotel?

In a real hotel, the system could be integrated with the hotel's POS system, inventory management software, and procurement system.

The workflow would be:

1. Daily consumption and inventory data would be collected automatically from the POS and inventory systems.

2. The system would clean and validate the data.

3. Forecasts would be generated for each beverage item and bar.

4. Recommended par levels and reorder quantities would be calculated.

5. The inventory manager would view alerts for low-stock and high-demand items through a dashboard.

6. Purchase recommendations would be reviewed and approved by the purchasing team.

7. Actual consumption and stockout information would be used to retrain and improve the model.

The system should initially operate as a decision-support tool, where hotel staff review recommendations before placing orders. After sufficient validation, it could be integrated with automated procurement workflows.

This would help the hotel improve inventory visibility, reduce stockout risks, minimize excess inventory, and make data-driven purchasing decisions.
