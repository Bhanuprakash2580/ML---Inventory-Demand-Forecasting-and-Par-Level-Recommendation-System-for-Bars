My project is an Inventory Demand Forecasting and Par Level Recommendation System for Bars.

The objective is to analyze historical inventory data, forecast future demand, and recommend suitable inventory levels.

I used Python, Pandas, NumPy, and forecasting techniques to preprocess the data and analyze consumption patterns.

I evaluated the forecasting results using MAE and WAPE metrics.

Next, I calculated recommended inventory par levels based on expected demand and safety stock to help reduce the risk of stockouts.

I developed an interactive Streamlit dashboard where users can explore the dataset, view consumption charts, check forecasts, and review inventory recommendations.
